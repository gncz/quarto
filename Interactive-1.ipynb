```python
q_sites_obs_range = """SELECT site_id
      ,site_name
      ,min(season_starting) AS "First observation year"
      ,max(season_starting) AS "Last observation year"
      ,max(season_starting) - min(season_starting) AS "Observation range in years"
      ,count(season_starting) AS "Number of observations"
      FROM observations 
      GROUP BY 
      site_id
      ,site_name;"""
      
df_sites_obs_range= pd.read_sql(q_sites_obs_range, conn)

df_sites_obs_range
```

```python
q_sites_obs_range = """SELECT site_id
      ,site_name
      ,min(season_starting) AS "First observation year"
      ,max(season_starting) AS "Last observation year"
      ,max(season_starting) - min(season_starting)+1 AS "Observation range in years"
      ,count(season_starting) AS "Number of observations"
      FROM observations 
      GROUP BY 
      site_id
      ,site_name;"""
      
df_sites_obs_range= pd.read_sql(q_sites_obs_range, conn)

df_sites_obs_range
```
